## Technical Steps

For this file the general plan for technical preprocessing was followed.

### Removal of Non-Informative Columns

First empty and duplicated columns were removed (see [](general:ecf)). Furthemore we removed the column `WNRegistrierungszentrumET`, because it contained encoded identifying information. We also removed `WNMinimumKreatininClearanceLetzterWertET` (therefore also `WNMinimumKreatininClearanceLetzterEinheitET`), as it always contained the same value or 0.

```python
data = drop_col_few_distinct(data)
data = drop_duplicate_columns(data).drop(
    columns=[
        "WNRegistrierungszentrumET",
        "WNMinimumKreatininClearanceLetzterWertET",
        "WNMinimumKreatininClearanceLetzterEinheitET",
    ]
)
```
