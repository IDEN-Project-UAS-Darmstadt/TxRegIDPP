## Technical Steps

For this file the general plan for technical preprocessing was followed.

### Removal of Non-Informative Columns

First empty were removed (see [](general:ecf)). We do not use the {term}`DSO` identifiers, so the column `SPostmLaborHLAIdDSOKennnummerDSO` was also removed. We furthermore dropped the column `SPostmLaborHLATypisierungszentrumET` which is an location identifier. Duplicate columns in HLA testing data might be important, therefore we did not remove them here.

```python
data = drop_col_few_distinct(data).drop(
    columns=["SPostmLaborHLATypisierungszentrumET", "SPostmLaborHLAIdDSOKennnummerDSO"]
)
# data = drop_duplicate_columns(data), entfernt SPostmLaborHLADRSP2ETDSO
```
