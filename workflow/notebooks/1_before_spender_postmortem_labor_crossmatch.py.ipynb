## Technical Steps

For this file the general plan for technical preprocessing was followed.

### Removal of Non-Informative Columns

First empty and duplicated columns were removed (see [](general:ecf)). We do not use the {term}`DSO` identifiers, so the column `SPostmLaborCrossmIdDSOKennnummerDSO` was also removed. We furthermore dropped the column `SPostmLaborCrossmTXCETDSO` which is an identifier.

```python
data = drop_col_few_distinct(data)
data = drop_duplicate_columns(data).drop(
    columns=["SPostmLaborCrossmIdDSOKennnummerDSO", "SPostmLaborCrossmTXCETDSO"]
)
```
