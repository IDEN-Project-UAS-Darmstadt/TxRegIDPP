# followup_niere

This file provides information on the follow-up visits with the recipient after the transplantation. Data is provided by {term}`ET` and {term}`IQTIG` in the file `element_followup_niere.csv`. The {term}`IQTIG` institute only provides a reference to the recipient which made filtering more difficult and needs to be considered when analyzing this dataset. 

We perform the common steps outlined in [](../02_preprocessing/index.md). The creatin units were converted and common translations applied and shared columns consolidated. 

## Unprocessed input data

```python
data = pd.read_csv(input_data, sep=";", low_memory=False)
display_data_doc(data=data, official_doc=pd.read_csv(input_doc))
```

## Technical Steps

For this file the general plan for technical preprocessing was followed.

### Removal of Non-Informative Columns

First empty and duplicated columns were removed (see [](general:ecf)).
