# spender_postmortem_untersuchungen

General examination results for deceased donors are provided by {term}`ET` and {term}`DSO` in the file `element_spender_postmortem_untersuchungen.csv`. 

We perform the common steps outlined in [](../02_preprocessing/index.md).

## Unprocessed input data

```python
data = pd.read_csv(input_data, sep=";", low_memory=False)
display_data_doc(data=data, official_doc=pd.read_csv(input_doc))
```

## Technical Steps

For this file the general plan for technical preprocessing was followed.

### Removal of Non-Informative Columns

First empty and duplicated columns were removed (see [](general:ecf)). We do not use the {term}`DSO` identifiers, so the column `SPostmUntersuchungIdDSOKennnummerDSO` was also removed.

```python
data = drop_col_few_distinct(data)
data = drop_duplicate_columns(data).drop(columns="SPostmUntersuchungIdDSOKennnummerDSO")
```

### Renaming of Columns

New names were used for the columns. (see [](general:cr)).

```python
renaming = {
    "SPostmUntersuchungBefundDateDSO": "result_date",
    "SPostmUntersuchungBefundergebnisDSO": "examination_result",
    "SPostmUntersuchungDateDSO": "examination_date",
    "SPostmUntersuchungDateET": "examination_date_et",
    "SPostmUntersuchungErfahrenAmDateDSO": "communicated_date",
    "SPostmUntersuchungIdSpenderNrETDSO": "donor_et_dso",
    "SPostmUntersuchungIdSpenderNrETET": "donor_et_id_et",
    "SPostmUntersuchungUntersuchungsverfahrenDSO": "examination_type",
    "SPostmUntersuchungUntersuchungsverfahrenET": "type_et",
}

data = rename(data, renaming)
data.to_parquet(output_data)
```
