Most of the data is provided by {term}`DSO`.

## Technical Steps

For this file the general plan for technical preprocessing was followed.

### Removal of Non-Informative Columns

First empty and duplicated columns were removed (see [](general:ecf)). We furthermore removed the columns `SPostmAnamnRaucherPYDSO` and `SPostmAnamnRaucherBeginnDSO` as they contained unstructured data compared to the same columns from {term}`ET` and the columns `SPostmBasisMeldungKHOrtDSO`, `SPostmBasisMeldungKHDSO`, `SPostmBasisDSORegionDSO` and `SPostmBasisMeldungKHOrtDSO` as those are location identifiers. We do not use the {term}`DSO` identifiers, so the column `SPostmIdDSOKennnummerDSO` was also removed. `SPostmBasisAbschaltungLebenserhaltungDateET` only contained invalid values.

```python
data = drop_col_few_distinct(data)
data = drop_duplicate_columns(data)
data.drop(
    [
        "SPostmAnamnRaucherPYDSO",
        "SPostmBasisMeldungKHOrtDSO",
        "SPostmBasisMeldungKHDSO",
        "SPostmBasisMeldungKHOrtDSO",
        "SPostmIdDSOKennnummerDSO",
        "SPostmBasisDSORegionDSO",
        "SPostmAnamnRaucherBeginnDSO",
        "SPostmBasisAbschaltungLebenserhaltungDateET",
    ],
    axis=1,
    inplace=True,
)
```
