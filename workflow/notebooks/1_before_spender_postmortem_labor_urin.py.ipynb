# spender_postmortem_labor_urin

Urine testing data is provided by {term}`ET` and {term}`DSO` in the file `element_spender_postmortem_labor_urin.csv`. 

We perform the common steps outlined in [](../02_preprocessing/index.md) and drop the {term}`ET` data as it only provides a sampling date.

## Unprocessed input data

```python
data = pd.read_csv(input_data, sep=";", low_memory=False)
display_data_doc(data=data, official_doc=pd.read_csv(input_doc))
```

## Technical Steps

For this file the general plan for technical preprocessing was followed.

### Removal of Non-Informative Columns

First empty and duplicated columns were removed (see [](general:ecf)). We do not use the {term}`DSO` identifiers, so the column `SPostmLaborURIIdDSOKennnummerDSO` was also removed.

```python
data = drop_col_few_distinct(data)
data = drop_duplicate_columns(data).drop(columns="SPostmLaborURIIdDSOKennnummerDSO")
```

### Renaming of Columns

New names were used for the columns. (see [](general:cr)).

```python
renaming = {
    "SPostmLaborURIBakterienDSO": "bacteria_in_urine",
    "SPostmLaborURIBefundDateDSO": "result_date",
    "SPostmLaborURIEpitheliumDSO": "epithelium_in_urine",
    "SPostmLaborURIErfahrenAmDateDSO": "communicated_date",
    "SPostmLaborURIErythrozytenDSO": "erythocytes_in_urine",
    "SPostmLaborURIGlukoseDSO": "glucose_in_urine",
    "SPostmLaborURIHBDSO": "hemoglobin_in_urine",
    "SPostmLaborURIIdSpenderNrETDSO": "donor_et_dso",
    "SPostmLaborURIIdSpenderNrETET": "donor_et_id_et",
    "SPostmLaborURIKetonkörperDSO": "ketone_in_urine",
    "SPostmLaborURILeukozytenDSO": "leukocytes_in_urine",
    "SPostmLaborURINitritDSO": "nitrite_in_urine",
    "SPostmLaborURIProbeDateDSO": "sampling_date_dso",
    "SPostmLaborURIProbeDateET": "sampling_date_et",
    "SPostmLaborURIProbenmaterialDSO": "sampling_tissue",
    "SPostmLaborURIProteinDSO": "protein_in_urine",
    "SPostmLaborURIUntersuchungDateDSO": "examination_date",
    "SPostmLaborURIZylinderDSO": "urinary_sediment",
    "SPostmLaborURIpHDSO": "urine_ph",
}

data = rename(data, renaming)
data.to_parquet(output_data)
```
